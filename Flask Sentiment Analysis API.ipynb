```python
from flask import Flask, jsonify, request
from transformers import pipeline

# declare flask and define app using Flask
app = Flask("sentiment_analysis")

tweets = {"tweets": [
                    {'content': 'dont know what todo',
                      'date': '17-02-2024',
                      'id': '1'},
                    {'content': 'doing flask',
                      'date': '16-03-2024',
                      'id': '2'}
                    ]
        }


@app.route("/")
def hello():
    return "Hello!"

@app.route('/tweet', methods=['GET'])
def return_all_tweets():
    return jsonify(tweets)

@app.route('/tweet/<string:date>', methods=["GET", "POST"])
def return_filtered_tweets_with_dates(date):
    if request.method == "POST":
        date = request.json["date"]
    tweet = [t for t in tweets["tweets"] if t["date"] == date]
    return jsonify({'data': tweet})

@app.route('/add_tweet', methods=["GET", 'POST'])
def add_tweet():
    tweet = {
                "content": request.json["content"],
                "date": request.json["date"],
                "id": request.json["id"]
            }
    tweets["tweets"].append(tweet)
    return jsonify(tweets)

@app.route('/classify_tweet', methods=['POST'])
def classify_tweet():
    sentiment_pipe = pipeline("sentiment-analysis",
                              model="siebert/sentiment-roberta-large-english")
    return jsonify(sentiment_pipe(request.json["content"]))

@app.route('/date', methods=['GET', 'POST'])
def get_tweets_by_date():
    if request.method == 'GET':
        date = request.args.get('date')
    else:
        date = request.json.get('date')
    filtered_tweets = [tweet for tweet in tweets["tweets"] if tweet['date'] == date]
    return jsonify({"tweets": filtered_tweets})

if __name__ == '__main__':
    app.run(host='0.0.0.0', port=5006)
```
